```python
import tensorflow as tf
from tensorflow.keras import layers, models
import matplotlib.pyplot as plt
import numpy as np

# 1. Load and Preprocess Data
print("Loading MNIST data...")
(x_train, y_train), (x_test, y_test) = tf.keras.datasets.mnist.load_data()
x_train, x_test = x_train / 255.0, x_test / 255.0 # Scale pixels to 0-1

# 2. Build the Neural Network Architecture
model = models.Sequential([
    layers.Flatten(input_shape=(28, 28)),      # Flattens 28x28 image to 784 pixels
    layers.Dense(128, activation='relu'),      # Hidden layer with 128 neurons
    layers.Dropout(0.2),                       # Prevents overfitting
    layers.Dense(10, activation='softmax')     # Output layer for 10 digits
])

# 3. Compile the Model
model.compile(optimizer='adam',
              loss='sparse_categorical_crossentropy',
              metrics=['accuracy'])

# 4. Train the Model
print("Starting training...")
model.fit(x_train, y_train, epochs=5)

# 5. Evaluate and Save
test_loss, test_acc = model.evaluate(x_test, y_test, verbose=2)
print(f'\nTest accuracy: {test_acc*100:.2f}%')

# Save the model file for GitHub
model.save('digit_model.h5')

# 6. Test a Prediction
prediction = model.predict(x_test[:1])
print(f"Prediction for first test image: {np.argmax(prediction)}")
```
